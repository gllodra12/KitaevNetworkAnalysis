```python
import pathlib
import pickle

import matplotlib.image as mpimg
import matplotlib.pyplot as plt
import networkx as nx
import numpy as np
import rsmf

REPO = pathlib.Path().cwd()
formatter = rsmf.setup(r"\documentclass[amsmath,amssymb,twocolumn,aps,pra,floatfix,superscriptaddress]{revtex4-2}")
```

```python
def upload_data(correlation, metric, configuration, root_path):
    # Upload data
    path = root_path.format(
        repository=REPO,
        correlation=correlation,
        metric=metric,
        configuration=configuration
    )
    x, y = np.load(path+"x_values.npy"), np.load(path+"y_values.npy")

    return x, y
```

```python
DATA_PLOT_A1 = "{repository}/data/fidelity/{configuration}/"
def plot_transition(configuration, color=None):

    path = DATA_PLOT_A1.format(repository=REPO, configuration=configuration)

    # Upload data
    x, y = np.load(path+"x_values.npy"), np.load(path+"y_values.npy")

    #  Hamiltonian parameters
    sites = int(configuration.split("_")[0])
    delta = configuration.split("_")[2]
    
    # Plot data
    plt.plot(x, y, label="$\Delta$={}".format(delta), color=color, linewidth=1.71)
    plt.xlim(0.9, 3.1)
    plt.ylim(-0.05, 1.05)
    

    plt.text(1.5, 0.4, "Topological\n phase", 
    ha="center", va="center", 
    bbox=dict(boxstyle="round",
            ec="#252525", #252525
            fc="#cccccc",
            )
    )
    plt.text(2.5, 0.4, "Trivial\n phase",
    ha="center", va="center",
    bbox=dict(boxstyle="round",
            ec="#252525",
            fc="#cccccc", #969696
            )
    )
        

def compare_transition(configurations: list):
    #plt.figure(dpi=150, figsize=[7.5, 4])
    fig = formatter.figure(aspect_ratio=0.55)

    for configuration, color in configurations:
        # Plot data
        plot_transition(configuration, color=color)
    
    # Labeling plot
    plt.xlabel(r"$\mu$")
    plt.ylabel("Fidelity") # fontweight='bold',
    plt.legend()
    plt.tight_layout(pad=0.7)
    plt.fill_between(np.linspace(0.9, 2, 10), [-0.05]*10, [1.05]*10, color='#969696', alpha=0.6) #pink=#f4d7eegrey=#636363, green=#bae4b3, brown=9e8e68
    plt.grid()
```

```python
# Plot Fig. A1
compare_transition([
    ("14_1.0_0.5_0.0_True_0.01", "#e41a1c"),
    ("14_1.0_1.0_0.0_True_0.01", "#1a9641"),
    ("14_1.0_2.0_0.0_True_0.01", "#2c7bb6")
])
plt.savefig("./plots_paper/FigA1_fidelity.pdf")
```
